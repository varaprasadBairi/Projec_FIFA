```python
def convert_rating(val):
    if isinstance(val, str) and '+' in val:
        base, plus = val.split('+')
        return int(base) + int(plus)
    return val

cols_to_fix = ['ls','st','rs','lw','lf','cf','rf','rw','lam','cam','ram',
               'lm','lcm','cm','rcm','rm','lwb','ldm','cdm','rdm','rwb',
               'lb','lcb','cb','rcb','rb']

for col in cols_to_fix:
    df[col] = df[col].apply(convert_rating)
```

# Removing Outliers

```python
cols = ['Age','overall','value_eur','wage_eur',
        'height_cm','weight_kg',
        'pace','shooting','passing','dribbling','defending','physic']
```

```python
for col in cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df = df[(df[col] >= lower) & (df[col] <= upper)]
```

```python
from sklearn.cluster import KMeans
wcss=[]
for i in range(1,8):
  k= KMeans(n_clusters=i, random_state=42)
  k.fit(df)
  wcss.append(k.inertia_)
wcss
```

```python
plt.plot(range(1,8),wcss)
plt.title('The Elbow Method')
plt.xlabel('Number of Clusters')
plt.ylabel('WCSS')
```

```python
model = KMeans(n_clusters=3,random_state=222)
model.fit(df)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
import pickle

pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("kmeans", KMeans(n_clusters=3, random_state=42))
])

# Remove Labels column before training
X = df.drop("Labels", axis=1)

pipeline.fit(X)

pickle.dump(pipeline, open("fifa_kmeans_pipeline.pkl", "wb"))
```
